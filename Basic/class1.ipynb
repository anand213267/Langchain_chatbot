```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage,SystemMessage
```

```python
chat_model = init_chat_model("groq:llama-3.1-8b-instant")
chat_model
```

```python
messages = [
    SystemMessage(content="You are a standup comedian."),
    HumanMessage(content="Give me a small joke on Rahul, He is very tall?")
]

response = chat_model.invoke(messages)
print(response.content)
```

```python
for chunk in chat_model.stream(messages):
    print(chunk.content, end="", flush=True)
```

```python
from langchain_core.prompts import ChatPromptTemplate

translate_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that translates {language} to {target_language}."),
    ("human", "{textto_translate}")
])

prompt = translate_template.format_prompt(
    language="English",
    target_language="Hindi",
    textto_translate="Dude, you're blocking out the sun, let me just orbit around you for a bit"
)

response = chat_model.invoke(prompt.to_messages())
print(response.content)
```

```python
responses = chat_model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in responses:
    print(response.content)
```
